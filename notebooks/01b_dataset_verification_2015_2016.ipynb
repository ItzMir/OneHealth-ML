```python
verified = {}

for comp_name, (fname, vars_list) in COMPONENTS.items():
    path = os.path.join(DATA_PATH, fname)
    if not os.path.exists(path):
        print(f"❌ File not found: {fname}")
        continue

    print(f"\n--- {comp_name} ---")
    df = pd.read_sas(path)
    df["cycle"] = 0   # 2015-2016

    dataset_summary(df, comp_name)

    # Add cycle to variable list for verification
    verify_variables(df, vars_list + ["cycle"])

    missing_summary(df)
    
    # Save with _2015 suffix
    out_path = f"../data/interim/{comp_name}_2015_verified.csv"
    df.to_csv(out_path, index=False)
    verified[comp_name] = df
    print(f"Saved: {out_path}")

print("\nAll available components verified.")
```

```python
if verified:
    merge_order = list(COMPONENTS.keys())
    merged = verified[merge_order[0]].copy()

    for comp in merge_order[1:]:
        if comp in verified:
            merged = merged.merge(verified[comp], on=["SEQN", "cycle"], how="inner")
            print(f"Merged {comp}: {merged.shape}")

    print(f"\nFinal 2015 merged shape: {merged.shape}")
    print("Duplicate SEQN:", merged["SEQN"].duplicated().sum())
else:
    print("No components loaded.")
```
