## Merge Tests

```python
merged = demo.copy()

datasets = [
    ("BMX_J", bmx),
    ("GHB_J", ghb),
    ("GLU_J", glu),
    ("BPX_J", bpx),
    ("BIOPRO_J", biopro),
    ("DIQ_J", diq),
    ("KIQ_U_J", kiq),
    ("SMQ_J", smq),
    ("ALQ_J", alq),
    ("PAQ_J", paq),
    ("DBQ_J", dbq),
    ("MCQ_J", mcq),
    ("BPQ_J", bpq),
    ("CBC_J", cbc),
    ("HDL_J", hdl),
    ("TCHOL_J", tchol),
    ("TRIGLY_J", trigly),
    ("INS_J", ins),
    ("HSCRP_J", hscrp),
    ("ALB_CR_J", alb_cr),
    ("HSQ_J", hsq)
]

print("=" * 60)
print("MERGE PROGRESS")
print("=" * 60)

for name, df in datasets:
    merged = merged.merge(df, on="SEQN", how="inner")
    print(f"After merging {name}: {merged.shape}")

print()
print("=" * 60)
print("FINAL MERGED DATASET")
print("=" * 60)
print(f"Rows    : {merged.shape[0]}")
print(f"Columns : {merged.shape[1]}")
print()
print("Duplicate SEQN:", merged["SEQN"].duplicated().sum())
```
