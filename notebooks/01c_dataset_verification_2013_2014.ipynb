```python
import pandas as pd, os
smq = pd.read_sas(os.path.join(DATA_PATH, "SMQ_H.xpt"))
print(sorted(smq.columns.tolist()))
```

```python
verified = {}

for comp_name, (fname, vars_list) in COMPONENTS.items():
    path = os.path.join(DATA_PATH, fname)
    if not os.path.exists(path):
        print(f"❌ File not found: {fname}")
        continue

    print(f"\n--- {comp_name} ---")
    df = pd.read_sas(path)
    df["cycle"] = 0   # temporary; will be reassigned later

    dataset_summary(df, comp_name)
    verify_variables(df, vars_list + ["cycle"])
    missing_summary(df)

    out_path = f"../data/interim/{comp_name}_2013_verified.csv"
    df.to_csv(out_path, index=False)
    verified[comp_name] = df
    print(f"Saved: {out_path}")

print("\nAll available 2013-2014 components verified.")
```

```python
if verified:
    merged = verified["DEMO"].copy()
    for comp in list(COMPONENTS.keys())[1:]:
        if comp in verified:
            merged = merged.merge(verified[comp], on=["SEQN", "cycle"], how="inner")
            print(f"Merged {comp}: {merged.shape}")
    print(f"\nFinal 2013 merged shape: {merged.shape}")
    print("Duplicate SEQN+cycle:", merged[["SEQN", "cycle"]].duplicated().sum())
```
